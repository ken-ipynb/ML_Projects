```python
# featuring of dataset using TF-IDF/countvectorize
import pandas as pd
dataset = pd.read_csv(r"C:\Users\61005\OneDrive\Desktop\jupyter notebook\Netflix Dataset.csv")
dataset[['Title', 'Category', 'Director', 'Cast', 'Type', 'Description']].head()

```

```python
dataset['tags'] = (
    dataset['Category'].astype(str) + ' ' +
    dataset['Director'].astype(str) + ' ' +
    dataset['Cast'].astype(str) + ' ' +
    dataset['Type'].astype(str) + ' ' +
    dataset['Description'].astype(str)
)
```

```python
dataset[['Title', 'tags']].head()
```

```python
dataset[['Title', 'tags']].head().values
```

```python
# now applying countvectorize from skitlearn (for featuring our  data )
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
vectors = cv.fit_transform(dataset['tags']).toarray()
vectors.shape
```

```python
# applying pca for our recommendation system
from sklearn.decomposition import PCA
pca = PCA(n_components=1000)
vectors_pca = pca.fit_transform(vectors).astype("float32")
vectors_pca.shape
```

```python
# checking variance from our original data and  are accurate approx 74%
pca.explained_variance_ratio_.sum()
```

```python
# prepration of train_data before training the model

type_sets = dataset["Type"].apply(
    lambda x: set(i.strip() for i in x.split(","))
)

print(type_sets.head())
```

```python
# Creating pairs and calculating similarity

import random
import torch
import torch.nn.functional as F

pairs = []
similarities = []

num_pairs = 30000


for _ in range(num_pairs):

    # Selecting two random movies
    i = random.randint(0, len(vectors_pca) - 1)
    j = random.randint(0, len(vectors_pca) - 1)

    # Don't compare a movie with itself
    while i == j:
        j = random.randint(0, len(vectors_pca) - 1)

    # Getting movie PCA vectors
    movie_a = torch.tensor(
        vectors_pca[i],
        dtype=torch.float32
    )

    movie_b = torch.tensor(
        vectors_pca[j],
        dtype=torch.float32
    )

    # Calculating cosine similarity
    similarity = F.cosine_similarity(
        movie_a.unsqueeze(0),
        movie_b.unsqueeze(0)
    ).item()

    # Storing movie pair
    pairs.append((i, j))

    # Storing similarity
    similarities.append(similarity)


print("Total pairs:", len(pairs))
print("Highest similarity:", max(similarities))
print("Lowest similarity:", min(similarities))
```

```python
# Combining pairs and similarities

combined = list(
    zip(pairs, similarities)
)


# Sorting by cosine similarity

combined.sort(
    key=lambda x: x[1],
    reverse=True
)


print("Highest similarity:", combined[0][1])
print("Lowest similarity:", combined[-1][1])
```

```python
# Creating balanced pairs and labels

pairs = []
labels = []

num_balanced = 5000


# Creating similar pairs

for pair, similarity in combined[:num_balanced]:

    pairs.append(pair)
    labels.append(1)


# Creating dissimilar pairs

for pair, similarity in combined[-num_balanced:]:

    pairs.append(pair)
    labels.append(0)


print("Total pairs:", len(pairs))
print("Similar pairs:", labels.count(1))
print("Dissimilar pairs:", labels.count(0))
```

```python
# Train and testing data split

from sklearn.model_selection import train_test_split

train_pairs, test_pairs, train_labels, test_labels = train_test_split(
    pairs,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)


print("Training pairs:", len(train_pairs))
print("Testing pairs:", len(test_pairs))

print("\nTraining similar:", train_labels.count(1))
print("Training dissimilar:", train_labels.count(0))

print("\nTesting similar:", test_labels.count(1))
print("Testing dissimilar:", test_labels.count(0))
```

```python
# pair dataset creation of movies
import torch
from torch.utils.data import Dataset

class MoviePairDataset(Dataset):

    def __init__(self, pairs, labels, vectors):
        self.pairs = pairs
        self.labels = labels
        self.vectors = vectors

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, idx):
        # getting the movies indices
        movie_a_idx, movie_b_idx = self.pairs[idx]
     # retirveing movie pca vectors
        movie_a = self.vectors[movie_a_idx]
        movie_b = self.vectors[movie_b_idx]
# getting similarity labels
        label = self.labels[idx]

        return movie_a, movie_b, label

train_dataset = MoviePairDataset(
    train_pairs,
    train_labels,
    vectors_pca
)

test_dataset = MoviePairDataset(
    test_pairs,
    test_labels,
    vectors_pca
)

movie_a, movie_b, label = train_dataset[0]

print("Movie A shape:", movie_a.shape)
print("Movie B shape:", movie_b.shape)
print("Label:", label)
```

```python
# Dataset loader

from torch.utils.data import DataLoader

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False
)

movie_a, movie_b, batch_labels = next(iter(train_loader))

print("Movie A batch:", movie_a.shape)
print("Movie B batch:", movie_b.shape)
print("Labels batch:", batch_labels.shape)
```

```python
# creation of recomendationNN
import torch 
import torch.nn as nn
class RecommendationNN(nn.Module):
    def __init__(self,in_features,hidden1,hidden2,hidden3,out_features):
        super().__init__()
        self.fc1=nn.Linear(in_features,hidden1)
        self.relu= nn.ReLU()
        self.fc2=nn.Linear(hidden1,hidden2)
        self.fc3=nn.Linear(hidden2,hidden3)
        self.fc4=nn.Linear(hidden3,out_features)

    def forward(self,x):
        x= self.fc1(x)
        x= self.relu(x)
        
        x=self.fc2(x)
        x=self.relu(x)
        
        x= self.fc3(x)
        x= self.relu(x)
        x=self.fc4(x)

        return x

model = RecommendationNN(
    in_features=1000,
    hidden1=512,
    hidden2=256,
    hidden3=128,
    out_features=64)

#cosine embedding loss (loss for our nn)
criterion = torch.nn.CosineEmbeddingLoss()

# NAG optimizers 
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.001,
    momentum=0.9,
    nesterov=True
)


# RecommendationNN training

# Number of epochs
epochs = 10

for epoch in range(epochs):

    model.train()
    total_loss = 0

    for movie_a, movie_b, labels in train_loader:

        # Convert labels: 0 → -1, 1 → 1
        # for our cosine embedding
        loss_labels = labels.float() * 2 - 1

        # Forward pass
        embedding_a = model(movie_a)
        embedding_b = model(movie_b)

        # Calculate loss
        loss = criterion(
            embedding_a,
            embedding_b,
            loss_labels
        )

        # Clear old gradients
        optimizer.zero_grad()

        # Backpropagation
        loss.backward()

        # Update NN weights using NAG
        optimizer.step()

        # Add batch loss
        total_loss += loss.item()

    # Calculate average loss for this epoch
    average_loss = total_loss / len(train_loader)

    print("Epoch:", epoch + 1)
    print("Average Training Loss:", average_loss)

print("Training completed")
```

```python
# RecommendationNN testing

model.eval()

total_test_loss = 0

with torch.no_grad():

    for movie_a, movie_b, labels in test_loader:

        # Convert labels: 0 → -1, 1 → 1
        loss_labels = labels.float() * 2 - 1

        # Forward pass
        embedding_a = model(movie_a)
        embedding_b = model(movie_b)

        # Calculate cosine loss
        loss = criterion(
            embedding_a,
            embedding_b,
            loss_labels
        )

        # Add test loss
        total_test_loss += loss.item()
         


# Calculate average test loss
average_test_loss = total_test_loss / len(test_loader)

print("Epoch:", epoch + 1)
print("Average Testing Loss:", average_test_loss)
print("Testing completed")
```

```python
# Generating embeddings for all movies

model.eval()

with torch.no_grad():

    all_vectors = torch.tensor(
        vectors_pca,
        dtype=torch.float32
    )

    movie_embeddings = model(all_vectors)

print("Movie embeddings shape:", movie_embeddings.shape)
```

```python
# Main part of RecommendationNN where we interact

import torch
import torch.nn.functional as F

# Evaluation mode
model.eval()


movie_name = input("Enter a movie you like: ")


# Find the movie in dataset
matches = dataset[
    dataset["Title"].str.lower() == movie_name.lower()
]


if len(matches) == 0:

    print("Movie not found.")


else:

    # Getting movie index
    movie_index = matches.index[0]

    # Selecting movie embedding
    selected_embedding = movie_embeddings[movie_index]


    # Calculate cosine similarity with all movies
    similarities = F.cosine_similarity(
        selected_embedding.unsqueeze(0),
        movie_embeddings
    )


    # Sorting movies by similarity
    top_indices = torch.argsort(
        similarities,
        descending=True
    )


    print("\nRecommended Movies:")

    count = 0

    for index in top_indices:

        index = index.item()

        # Skip the selected movie itself
        if index == movie_index:
            continue

        print(
            dataset.loc[index, "Title"],
            "→ Similarity:",
            round(similarities[index].item(), 4)
        )

        count += 1

        if count == 5:
            break
```
