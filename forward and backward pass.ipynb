```python
# forward pass
# normal input static
import torch
import torch.nn as nn

#creating layers
fc1 = nn.Linear(2,3)
relu= nn.ReLU()
fc2=nn.Linear(3,1)
x=torch.tensor([[10.0,20.0]])
x=fc1(x)
print("after fc1")
print(x)
x=relu(x)
print("after relu")
print(x)
x=fc2(x)
print("final prediction")
print(x)
```

```python
# dynamic inputs/feature
import torch
import torch.nn as nn
fc1= nn.Linear(2,4)
relu=nn.ReLU()
fc2 = nn.Linear(4,1)
f1 = float(input("Enter Feature 1: "))
f2 = float(input("Enter Feature 2: "))

x= torch.Tensor([[f1,f2]])
x = fc1(x)
print("\nAfter fc1:")
print(x)

x = relu(x)
print("\nAfter ReLU:")
print(x)

x = fc2(x)
print("\nPrediction:")
print(x)
```

```python
# backward pass using static input/features
import torch


x = torch.tensor(4.0, requires_grad=True)


y = x ** 2

print("y =", y)


y.backward()


print("dy/dx =", x.grad)
```

```python
# Backward pass using dynamic features
import torch
f1 = float(input("Enter Feature 1: "))
f2= float(input("Enter Feature 1: "))
x=torch.tensor([f1,f2],requires_grad=True, dtype=torch.float32)
y = x[0]**3 + x[1]
print("y= ",y)
y.backward()
print("dy/dx =", x.grad)
```
